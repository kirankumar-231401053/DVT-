```python

import pandas as pd
df=pd.read_csv("ufc_gold_dataset_final.csv")
df.head()
```

```python
# EXPERIMENT 4
# EDA – DATA INSPECTION, FILTERING, STATISTICAL ANALYSIS AND CORRELATION

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt



# --------------------------------------------------
# PROGRAM 2: Shape of Dataset
# --------------------------------------------------

print("Shape of Dataset:", df.shape)


# --------------------------------------------------
# PROGRAM 3: Data Types
# --------------------------------------------------

print("\nData Types:")
print(df.dtypes)


# --------------------------------------------------
# PROGRAM 4: Filtering – Fights with more than 5
# Significant Strikes by Fighter 1
# --------------------------------------------------

high_strike_fights = df[df["F1_Sig_Landed"] > 5]

print("\nFights where Fighter 1 landed more than 5 significant strikes:")
display(
    high_strike_fights[
        ["Fighter_1", "Fighter_2", "Winner", "F1_Sig_Landed"]
    ].head(10)
)


# --------------------------------------------------
# PROGRAM 5: Filtering – Fights from a Weight Class
# --------------------------------------------------

print("\nWeight Classes:")
print(df["Weight_Class"].value_counts().head(10))

# Select the most common weight class automatically
common_weight_class = df["Weight_Class"].value_counts().index[0]

weight_class_fights = df[
    df["Weight_Class"] == common_weight_class
]

print(f"\nFights in {common_weight_class}:")
display(
    weight_class_fights[
        ["Fighter_1", "Fighter_2", "Winner", "Weight_Class"]
    ].head(10)
)


# --------------------------------------------------
# PROGRAM 6: Sorting
# Sort fights by Fighter 1 significant strikes
# in descending order
# --------------------------------------------------

sorted_df = df.sort_values(
    by="F1_Sig_Landed",
    ascending=False
)

print("\nTop fights by Fighter 1 significant strikes landed:")
display(
    sorted_df[
        ["Fighter_1", "Fighter_2", "Winner", "F1_Sig_Landed"]
    ].head(10)
)


# --------------------------------------------------
# PROGRAM 7: Statistical Analysis
# --------------------------------------------------

print("\nMean Fighter 1 Significant Strikes =",
      df["F1_Sig_Landed"].mean())

print("Median Fighter 1 Significant Strikes =",
      df["F1_Sig_Landed"].median())

print("Mode Fighter 1 Significant Strikes =",
      df["F1_Sig_Landed"].mode().iloc[0])

print("Standard Deviation =",
      df["F1_Sig_Landed"].std())


# --------------------------------------------------
# PROGRAM 8: Descriptive Statistics
# --------------------------------------------------

print("\nDescriptive Statistics:")
display(
    df[
        [
            "F1_Sig_Landed",
            "F2_Sig_Landed",
            "F1_KD",
            "F2_KD",
            "Total_Fight_Time_Sec"
        ]
    ].describe()
)


# --------------------------------------------------
# PROGRAM 9: Grouping and Aggregation
# Average significant strikes by Weight Class
# --------------------------------------------------

weight_avg = df.groupby(
    "Weight_Class"
)["F1_Sig_Landed"].mean().sort_values(
    ascending=False
)

print("\nAverage Fighter 1 Significant Strikes by Weight Class:")
display(weight_avg)


# Visualization – Average Significant Strikes
plt.figure(figsize=(12, 6))
weight_avg.head(10).plot(kind="bar")

plt.title("Average Significant Strikes by Weight Class")
plt.xlabel("Weight Class")
plt.ylabel("Average Significant Strikes")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# --------------------------------------------------
# PROGRAM 10: Correlation Analysis
# --------------------------------------------------

corr = df[
    [
        "F1_Sig_Landed",
        "F2_Sig_Landed",
        "F1_KD",
        "F2_KD",
        "Total_Fight_Time_Sec"
    ]
].corr()

print("\nCorrelation Matrix:")
display(corr)


# Correlation Heatmap
plt.figure(figsize=(8, 6))

plt.imshow(corr, interpolation="nearest")
plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("UFC Fight Statistics Correlation")
plt.tight_layout()
plt.show()
```
