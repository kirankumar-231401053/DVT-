```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('Player.csv')

print("Dataset Loaded Successfully")
df.head()
```

```python
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
print("Columns :", df.columns.tolist())
```

```python
df.dtypes
```

```python
df['birthday'] = pd.to_datetime(df['birthday'])

df['Year'] = df['birthday'].dt.year
df['Month'] = df['birthday'].dt.month
df['Day'] = df['birthday'].dt.day
df['Hour'] = df['birthday'].dt.hour

df.head()
```

```python
player_names = df['player_name'].head(6)
height = df['height'].head(6)
weight = df['weight'].head(6)
```

```python
# Program 1: Line Chart
plt.plot(player_names, height)
plt.title("Player Height")
plt.xlabel("Players")
plt.ylabel("Height")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Program 2: Bar Chart
plt.bar(player_names, height)
plt.title("Player Height")
plt.xlabel("Players")
plt.ylabel("Height")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Program 3: Pie Chart
plt.pie(height, labels=player_names, autopct="%1.1f%%")
plt.title("Player Height Distribution")
plt.tight_layout()
plt.show()
```

```python
# Program 4: Histogram
plt.hist(df['height'].dropna(), bins=10)
plt.title("Player Height Distribution")
plt.xlabel("Height")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Program 5: Scatter Plot
plt.scatter(df['height'], df['weight'])
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Player Height vs Weight")
plt.tight_layout()
plt.show()
```

```python
# Program 6: Box Plot
plt.boxplot(df['weight'].dropna())
plt.title("Player Weight Box Plot")
plt.ylabel("Weight")
plt.show()
```

```python
# Program 7: Multiple Line Charts
plt.plot(player_names, height, label="Height")
plt.plot(player_names, weight, label="Weight")
plt.title("Player Height and Weight")
plt.xlabel("Players")
plt.ylabel("Value")
plt.xticks(rotation=45, ha="right")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Player dataset
df = pd.read_csv("Player.csv")

# Display dataset
print("Dataset Preview:")
display(df.head())

# Select numerical columns
num_cols = df.select_dtypes(include="number").columns
print("\nNumerical columns:")
print(num_cols.tolist())
```

```python
# 1. BAR CHART
top10 = df.sort_values(
    "weight",
    ascending=False
).head(10)

plt.figure(figsize=(10,5))
plt.bar(
    top10["player_name"],
    top10["weight"]
)
plt.title("Player Dataset - Bar Chart")
plt.xlabel("Player")
plt.ylabel("Weight")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# 2. LINE CHART
first20 = df.head(20)

plt.figure(figsize=(9,5))
plt.plot(
    range(1, len(first20) + 1),
    first20["weight"],
    marker="o"
)
plt.title("Player Dataset - Line Chart")
plt.xlabel("Player Record")
plt.ylabel("Weight")
plt.grid()
plt.tight_layout()
plt.show()
```

```python
# 3. PIE CHART
plt.figure(figsize=(7,7))
plt.pie(
    top10["weight"],
    labels=top10["player_name"],
    autopct="%1.1f%%"
)
plt.title("Player Dataset - Top 10 Weight Share")
plt.tight_layout()
plt.show()
```

```python
# 4. BOX PLOT
plt.figure(figsize=(8,5))
sns.boxplot(y=df["weight"])
plt.title("Player Dataset - Box Plot")
plt.ylabel("Weight")
plt.tight_layout()
plt.show()
```

```python
# 5. HISTOGRAM
plt.figure(figsize=(8,5))
plt.hist(
    df["weight"],
    bins=15
)
plt.title("Player Dataset - Histogram")
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
