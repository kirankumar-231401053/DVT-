```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('Player.csv')

print("Dataset Loaded Successfully")
df.head()
```

```python
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
print("Columns :", df.columns.tolist())
```

```python
df.dtypes
```

```python
df['birthday'] = pd.to_datetime(df['birthday'])

df['Year'] = df['birthday'].dt.year
df['Month'] = df['birthday'].dt.month
df['Day'] = df['birthday'].dt.day
df['Hour'] = df['birthday'].dt.hour

df.head()
```

```python
df.describe(include='all')
```

```python
print(df.select_dtypes(include='number').columns.tolist())
```

```python
print(df.select_dtypes(include='object').columns.tolist())
```

```python
player_data = df[df['height'] > 180]

player_data.head()
```

```python
sorted_df = df.sort_values(by='weight', ascending=False)

sorted_df.head()
```

```python
column = 'weight'

print("Mean :", df[column].mean())
print("Median :", df[column].median())
print("Mode :")
print(df[column].mode())

print("Standard Deviation :", df[column].std())
```

```python
grouped = df.groupby('Year')['weight'].mean()

print(grouped)
```

```python
grouped.head(10).plot(kind='bar', figsize=(8,5))

plt.title('Average Weight by Birth Year')
plt.xlabel('Birth Year')
plt.ylabel('Average Weight')

plt.show()
```

```python
corr = df.select_dtypes(include=np.number).corr()

print(corr)
```

```python
plt.figure(figsize=(8,6))

plt.imshow(corr, cmap='coolwarm', interpolation='nearest')

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title('Correlation Matrix')

plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.scatter(df['height'], df['weight'])

plt.xlabel('Height')
plt.ylabel('Weight')

plt.title('Scatter Plot')

plt.show()
```
