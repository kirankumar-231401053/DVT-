```python

import pandas as pd
df=pd.read_csv("ufc_gold_dataset_final.csv")
df.head()
```

```python
# EXPERIMENT 5
# DATA VISUALIZATION USING MATPLOTLIB – UFC DATASET

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt



print("Dataset loaded successfully!")
print("Shape:", df.shape)


# ==========================================================
# 1. LINE CHART
# Number of UFC fights by year
# ==========================================================

df["Event_Date"] = pd.to_datetime(df["Event_Date"], errors="coerce")

fights_per_year = df["Event_Date"].dt.year.value_counts().sort_index()

plt.figure(figsize=(10, 5))
plt.plot(fights_per_year.index, fights_per_year.values, marker="o")

plt.title("Number of UFC Fights by Year")
plt.xlabel("Year")
plt.ylabel("Number of Fights")
plt.grid(True)
plt.tight_layout()
plt.show()


# ==========================================================
# 2. BAR CHART
# Top 10 Weight Classes by Number of Fights
# ==========================================================

weight_class_counts = df["Weight_Class"].value_counts().head(10)

plt.figure(figsize=(10, 6))
plt.bar(weight_class_counts.index, weight_class_counts.values)

plt.title("Top 10 UFC Weight Classes by Number of Fights")
plt.xlabel("Weight Class")
plt.ylabel("Number of Fights")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# ==========================================================
# 3. PIE CHART
# Distribution of Fight Results
# ==========================================================

result_counts = df["Winner"].value_counts().head(3)

plt.figure(figsize=(7, 7))
plt.pie(
    result_counts.values,
    labels=result_counts.index,
    autopct="%1.1f%%"
)

plt.title("UFC Fight Result Distribution")
plt.tight_layout()
plt.show()


# ==========================================================
# 4. HISTOGRAM
# Distribution of Total Fight Time
# ==========================================================

plt.figure(figsize=(10, 5))
plt.hist(
    df["Total_Fight_Time_Sec"],
    bins=30
)

plt.title("Distribution of UFC Fight Duration")
plt.xlabel("Fight Duration (Seconds)")
plt.ylabel("Number of Fights")
plt.tight_layout()
plt.show()


# ==========================================================
# 5. SCATTER PLOT
# Significant Strikes vs Fight Duration
# ==========================================================

plt.figure(figsize=(10, 6))

plt.scatter(
    df["Total_Fight_Time_Sec"],
    df["F1_Sig_Landed"],
    alpha=0.4
)

plt.title("Fight Duration vs Fighter 1 Significant Strikes")
plt.xlabel("Fight Duration (Seconds)")
plt.ylabel("Fighter 1 Significant Strikes Landed")
plt.tight_layout()
plt.show()


# ==========================================================
# 6. BOX PLOT
# Fighter 1 vs Fighter 2 Significant Strikes
# ==========================================================

plt.figure(figsize=(8, 6))

plt.boxplot(
    [
        df["F1_Sig_Landed"],
        df["F2_Sig_Landed"]
    ],
    labels=["Fighter 1", "Fighter 2"]
)

plt.title("Distribution of Significant Strikes")
plt.xlabel("Fighter")
plt.ylabel("Significant Strikes Landed")
plt.tight_layout()
plt.show()


# ==========================================================
# SUMMARY
# ==========================================================

print("\nVisualization completed successfully!")
print("Charts created:")
print("1. Line Chart - UFC Fights by Year")
print("2. Bar Chart - Weight Class Distribution")
print("3. Pie Chart - Fight Result Distribution")
print("4. Histogram - Fight Duration")
print("5. Scatter Plot - Fight Duration vs Significant Strikes")
print("6. Box Plot - Fighter 1 vs Fighter 2 Significant Strikes")
```
