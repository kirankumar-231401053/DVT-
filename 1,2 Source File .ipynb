```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv("Player.csv")
print("Dataset Loaded Successfully!")
```

```python
df.head()
```

```python
df.tail()
```

```python
print(df.columns.tolist())
```

```python
print("Duplicate Records:", df.duplicated().sum())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df.to_excel("Player.xlsx", index=False)

print("Excel file created successfully!")
```

```python
excel_data = pd.read_excel("Player.xlsx")

excel_data.head()
```

```python
conn = sqlite3.connect("Player.db")

df.to_sql("Player", conn, if_exists="replace", index=False)

print("Database created successfully!")
```

```python
query = "SELECT * FROM Player"

sql_data = pd.read_sql(query, conn)

sql_data.head()
```

```python
df.to_csv("Player_Output.csv", index=False)

print("CSV file exported successfully!")
```

```python
df.to_excel("Player_Output.xlsx", index=False)

print("Excel file exported successfully!")
```

```python
df.to_json("Player_Output.json", orient="records", indent=4)

print("JSON file exported successfully!")
```

```python
print("CSV File:")
print(pd.read_csv("Player_Output.csv").head())

print("\nExcel File:")
print(pd.read_excel("Player_Output.xlsx").head())

print("\nJSON File:")
print(pd.read_json("Player_Output.json").head())
```
